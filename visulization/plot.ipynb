```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Data
df = pd.DataFrame({
    'Age': [22,25,30,35,40,45,50,55],
    'Salary': [25000,30000,45000,50000,65000,70000,85000,90000],
    'Department': ['IT','HR','IT','Finance','HR','IT','Finance','HR']
})
```

```python
# !pip install seaborn
```

1. Histogram

```python
plt.hist(df['Age'], bins=5)
plt.title("Histogram")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

2. Bar Plot


```python
sns.barplot(x='Department', y='Salary', data=df)
plt.title("Bar Plot")
plt.show()
```

3. Pie Chart


```python
df['Department'].value_counts().plot.pie(
    autopct='%1.1f%%'
)
plt.title("Pie Chart")
plt.ylabel("")
plt.show()
```

4. Line Plot


```python
plt.plot(df['Age'], df['Salary'])
plt.title("Line Plot")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

5. Scatter Plot


```python
plt.scatter(df['Age'], df['Salary'])
plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

6. Box Plot


```python
sns.boxplot(y=df['Salary'])
plt.title("Box Plot")
plt.show()
```

8. KDE Plot


```python
sns.kdeplot(df['Salary'], fill=True)
plt.title("KDE Plot")
plt.show()
```

9. Count Plot


```python
sns.countplot(x='Department', data=df)
plt.title("Count Plot")
plt.show()
```

10. Heatmap


```python
corr = df[['Age','Salary']].corr()

sns.heatmap(corr,
            annot=True)
plt.title("Heatmap")
plt.show()
```

11. Pair Plot


```python
sns.pairplot(df)
plt.show()
```

3. Confusion Matrix


```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

y_true = [0,1,1,0,1]
y_pred = [0,1,0,0,1]

cm = confusion_matrix(y_true, y_pred)

ConfusionMatrixDisplay(cm).plot()
plt.show()
```

14. ROC Curve


```python
from sklearn.metrics import roc_curve

y_true = [0,0,1,1]
y_score = [0.1,0.4,0.35,0.8]

fpr, tpr, _ = roc_curve(y_true, y_score)

plt.plot(fpr, tpr)
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curve")
plt.show()
```

# Scatter Plot with Regression Line


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Data
df = pd.DataFrame({
    'Age': [22,25,30,35,40,45,50,55],
    'Salary': [25000,30000,45000,50000,65000,70000,85000,90000]
})

# Scatter Plot with Regression Line
sns.regplot(
    x='Age',
    y='Salary',
    data=df
)

plt.title("Scatter Plot with Regression Line")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
