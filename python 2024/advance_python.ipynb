# getter and setter method

```python
class Mobile:
    
    def __init__(self) -> None:
        self.name = ""
        
    @property   
    def get_mobile(self):
        return self.name
        
    @get_mobile.setter
    def set_mobile(self,value):
        self.name = value
```

```python
m = Mobile()
m.get_mobile
m.set_mobile = "ddd"
m.get_mobile
```

## static Method

```python
class Mobile:
    fp = "Yes"
    
    @staticmethod
    def myshow():
        print(Mobile.fp)
        
m = Mobile()
m.myshow()
Mobile.myshow()
```

# Inheritance

```python
class Father:
    def money(self):
        print("Father Class ---- Instance method")
    
    @classmethod
    def show(cls):
        print("Father Class --- this is class method")
        
    @staticmethod
    def displ():
        print("Father class - This is static method")
        
class Son(Father):
    def mybal(self):
        print("Son class Instance method mybal")
        
        
s = Son()
s.mybal()
s.displ()
s.show()
s.money()
```

# constructor overriding

```python
class Father:
    def __init__(self,r) -> None:
        print("Father class constructor")
        self.monoey = r
        
class Son(Father):
    
    def __init__(self,r) -> None:
        print("Son class constructor")
        self.monoey = r
    
    def disp(self):
        print("son class ==> display method --> ")
        
    
s = Son(10000)
s.monoey
```

# Single Inharitence

```python
class Father:
    def __init__(self,r) -> None:
        print("Father class constructor")
        self.monoey = r
        
class Son(Father):
    
    def __init__(self,r) -> None:
        print("Son class constructor")
        super().__init__(r)
        
    
    def disp(self):
        print("son class ==> display method --> ")
        
    
s = Son(10000)
print("father class" , s.monoey)
```

#  Multi-Level Constructor

```python
class Father:
    def __init__(self) -> None:
        print("Fater class constructor")
        
class Son(Father):
    
    def __init__(self) -> None:
        super().__init__()
        print("Son class constructor")
          
class GrandSon(Son):
    def __init__(self) -> None:
        super().__init__()
        print("GrandSon class constructor")
    
g = GrandSon()
```

# Herachical Constructor

```python
class Father:
    def __init__(self) -> None:
        print("Fater class constructor")
        
class Son(Father):
    
    def __init__(self) -> None:
        super().__init__()
        print("Son class constructor")
          
        
class Doughter(Father):
    
    def __init__(self) -> None:
        super().__init__()
        print("Doughter class constructor")
        
        
s = Son()
d = Doughter()
```

# Multi-Level Constructor

```python
class Father:
    def __init__(self) -> None:
        super().__init__()
        print("Father class constructor")
        
class Mother:
    def __init__(self) -> None:
        super().__init__()
        print("Mother class constructor")
          
        
class Child(Father,Mother):
    def __init__(self) -> None:
        super().__init__()
        print("Child class constructor")
        
        
s = Child()
```
