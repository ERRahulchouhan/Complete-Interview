```python
class Node:
    def __init__(self,data):
        self.value = data
        self.next = None
```

```python
class SlinkList:
    
    def __init__(self):
        self.head = None
        
    def insert(self,data):
        new_node = Node(data)
        
        if self.head is None:
            self.head = new_node
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
        
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")
        

# Example
sll = SlinkList()
sll.insert(10)
sll.insert(20)
sll.insert(30)  
sll.display()      
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")

    def reverse(self):
        prev = None
        current = self.head
        self.tail = self.head

        while current:
            nxt = current.next
            current.next = prev
            prev = current
            current = nxt

        self.head = prev


# Example
ll = LinkedList()

ll.insert(10)
ll.insert(20)
ll.insert(30)

print("Before Reverse:")
ll.display()

ll.reverse()

print("After Reverse:")
ll.display()
```

2. Doubly Linked List Reverse (Head + Tail)

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.prev = None
        self.next = None

class DLL:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert(self,data):
        new_node = Node(data)

        if self.head is None:
            self.head = self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

    def display(self):
        temp = self.head

        while temp:
            print(temp.data,end=" <-> ")
            temp = temp.next

        print("None")

    def reverse(self):
        current = self.head

        while current:
            current.prev, current.next = current.next, current.prev
            current = current.prev

        self.head, self.tail = self.tail, self.head


dll = DLL()

dll.insert(10)
dll.insert(20)
dll.insert(30)

print("Before Reverse:")
dll.display()

dll.reverse()

print("After Reverse:")
dll.display()
```

3. Circular Linked List Reverse (Head + Tail)

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

class CLL:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert(self,data):
        new_node = Node(data)

        if self.head is None:
            self.head = self.tail = new_node
            self.tail.next = self.head
        else:
            self.tail.next = new_node
            self.tail = new_node
            self.tail.next = self.head

    def display(self):
        temp = self.head

        while True:
            print(temp.data,end=" -> ")
            temp = temp.next

            if temp == self.head:
                break

        print("(Head)")

    def reverse(self):
        prev = self.tail
        current = self.head

        while True:
            nxt = current.next
            current.next = prev

            prev = current
            current = nxt

            if current == self.head:
                break

        self.head, self.tail = self.tail, self.head


cll = CLL()

cll.insert(10)
cll.insert(20)
cll.insert(30)

print("Before Reverse:")
cll.display()

cll.reverse()

print("After Reverse:")
cll.display()
```
